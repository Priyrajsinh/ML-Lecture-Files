```python
from sklearn.model_selection import train_test_split
```

```python
X = df[['Avg. Session Length', 'Time on App','Time on Website', 'Length of Membership']]
y = df[['Yearly Amount Spent']]
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

### Traing the Model:

```python
from sklearn.linear_model import LinearRegression
lm = LinearRegression()
```

```python
lm.fit(X_train, y_train)
```

```python
cdf = lm.coef_
```

```python
# Flatten coefficients to 1D and align with feature names
#lm.coef_.ravel()
cdf = pd.DataFrame(lm.coef_.ravel(),X.columns, columns=['coef'])
print(cdf)
```

### Predictions:

```python
predictions = lm.predict(X_test)
```

```python
predictions
```

```python
#sns.scatterplot(x=predictions, y=y_test)
sns.scatterplot(x=np.ravel(predictions), y=np.ravel(y_test))
plt.xlabel("Predictions")
plt.title("Evaluation of our LM Model")
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import math
```

```python
print("Mean Absolute Error", mean_absolute_error(y_test,predictions))
print("Mean Squared Error", mean_squared_error(y_test,predictions))
print("Root Mean Squared Error", math.sqrt(mean_squared_error(y_test,predictions)))
```

### Residuals

```python
residuals = y_test - predictions
```
