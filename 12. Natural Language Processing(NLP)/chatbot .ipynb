```python
#  CHATBOT USING INTENT MATCHING

import nltk
nltk.download('punkt')  # punkt is a pre-trained tokenizer model in NLTK

def get_intent(user_input):
    text = user_input.lower()    #    HELLO -  hello

    # Greetings
    if any(word in text for word in ["hello", "hi", "hey", "hola"]):
        return "greet"

    # Who are you?
    if any(word in text for word in ["who are you", "your name", "tell me about you"]):
        return "identity"

    # What can you do?
    if any(word in text for word in ["what can you do", "help", "abilities", "skills"]):
        return "ability"

    # Exit
    if any(word in text for word in ["bye", "exit", "goodbye"]):
        return "exit"

    return "unknown"


def get_response(intent):
    responses = {
        "greet": "Hello Buddy! How can I help you?",
        "identity": "I am a simple chatbot created to talk with you.",
        "ability": "I can answer basic questions for you.",
        "exit": "Goodbye Buddy! Take care.",
        "unknown": "Sorry Buddy, I didn't understand that."
    }
    return responses[intent]


print("ChatBot: Hello Buddy! Type 'exit' anytime.")

while True:
    user_input = input("You: ")

    intent = get_intent(user_input)
    response = get_response(intent)

    print("ChatBot:", response)

    if intent == "exit":
        break

```
