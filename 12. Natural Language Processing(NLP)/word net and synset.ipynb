# What is WordNet?

WordNet is a lexical database of English words used in Natural Language Processing (NLP).

📌 Developed at Princeton University

👉 It groups words based on meaning, not spelling.

# Why WordNet is Needed?

In natural language:

One word can have many meanings

Many words can have same meaning

Example

Word: bank

River bank

Money bank

Computers cannot understand this without semantic knowledge.

✔ WordNet helps machines understand meaning and relationships between words


```python
import nltk  # NLTK (Natural Language Toolkit) library,Tokenization,Stemming,Lemmatization,WordNet access
from nltk.corpus import wordnet as wn  # Imports WordNet corpus from NLTK

nltk.download('wordnet')  # Downloads WordNet lexical database,Contains:Words,Synsets,Definitions,Semantic relations
nltk.download('omw-1.4')  # OMW = Open Multilingual WordNet, Word meanings across languages,Better lemmatization
```

```python
wn.synsets("bank")  # It finds all possible meanings (synsets) of the word "bank" from WordNet
```

```python
for syn in wn.synsets("bank"):  # wn.synsets("bank") → returns all synsets (meanings) of the word bank,syn → one synset at a time
    print(syn.name())  # The synset identifier
    print("Definition:", syn.definition())  # syn.definition(),Returns the meaning of the current synset
    print("Example:", syn.examples())  # print("Definition:", ...),Displays that meaning on the screen
    print()
```

```python
dog = wn.synsets("dog")[0]  # wn.synsets("dog") → returns all meanings (synsets) of “dog”,[0] → selects the first and most common meaning

print("Hypernyms:", dog.hypernyms())   # IS-A  # Hypernym = IS-A relationship (general word)  # A hypernym is a more general word that answers, Hypernym = parent class
print("Hyponyms:", dog.hyponyms())     # Types  # Hyponym = Types or sub-categories  # Hyponym = child class
print("Meronyms:", dog.part_meronyms()) # Parts  # Meronym = Part-of relationship  # Meronym = component / part
# Tail is part of dog
#Paw is part of dog
#Snout is part of dog
```

# What is a Synset?

🔹 Synset = Synonym Set
* Synonym	Words with same meaning
Synset	A group of synonyms representing ONE meaning

A synset is a group of words with the same meaning.

📌 Each synset represents one concept

```python
{'car', 'auto', 'automobile'}
```

```python
synsets = wn.synsets("happy")  # finds all possible meanings (synsets) of the word “happy” from WordNet and stores them in the variable synsets
synsets
```

```python
syn = wn.synsets("happy")[0]  # wn.synsets("happy"), Returns all meanings (synsets) of the word happy,[0]→ Selects the first (most common) meaning
syn.lemma_names()  # lemma is the base form of a word, In this particular sense, these words share the same meaning
```

```python
happy = wn.synsets("happy")[0]  # wn.synsets("happy") returns all meanings (synsets) of the word happy,[0] selects the first and most common meaning

for lemma in happy.lemmas():  # A synset contains multiple lemmas,A lemma is an actual word (synonym) in that synset
    if lemma.antonyms():  # Checks whether the current lemma has an antonym,Some lemmas may not have antonyms, so this condition avoids errors, antonym is a word that has the opposite meaning of another word
        print(lemma.antonyms()[0].name())  # lemma.antonyms() returns a list of antonym lemmas,[0] selects the first antonym.name() prints the actual word
```

```python
dog = wn.synsets("dog")[0]  # Gets all meanings of dog,Selects the most common meaning
cat = wn.synsets("cat")[0]  # Gets all meanings of cat,Selects the most common meaning

dog.wup_similarity(cat)  # Wu–Palmer similarity score between the two synsets, Their lowest common ancestor (LCA) in the WordNet hierarchy, It tells how similar the meanings of two words are, based on their position in the WordNet hierarchy
```

```python
from collections import defaultdict  # Groups those meanings by noun, verb, adjective, adverb

meanings = defaultdict(list)  # Give me all meanings (synsets) of the word light

for syn in wn.synsets("light"):  # syn = one meaning at a time
    meanings[syn.pos()].append(syn.definition())  # syn.definition()This returns the dictionary meaning (definition) of that synset, Put this definition into the list that corresponds to its part of speech

for pos, defs in meanings.items():
    print(pos.upper())
    for d in defs:
        print(" -", d)
```
