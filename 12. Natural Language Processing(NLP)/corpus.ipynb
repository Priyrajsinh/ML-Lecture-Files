# What is a Corpus?

A corpus (plural: corpora) is a large, structured collection of text used to:

Train NLP models

Analyze language

Extract patterns, frequency, and meaning

📌 In simple words:

Corpus = raw text data for NLP

# Why Corpus is Important?

All NLP tasks depend on data:

Tokenization

POS tagging

Word embeddings

Language models

Chatbots

Translation

👉 No corpus = no NLP

# Types of Corpus
## 1️⃣ Raw Corpus

Plain text without labels
Example:

I love NLP.
* NLP is powerful.

## 2️⃣ Annotated Corpus

Text + linguistic labels

Example:

I/PRON love/VERB NLP/NOUN


Used for:

POS tagging

NER

Parsing

## 3️⃣ Monolingual Corpus

Single language
* Example: English news articles

## 4️⃣ Multilingual Corpus

Multiple languages
Example:

English ↔ Hindi


Used in machine translation

## 5️⃣ Specialized Corpus

Domain-specific text
Example:

Medical corpus

Legal corpus

```python
# Install & Import NLTK
import nltk
```

```python
# Download a Corpus  
nltk.download('brown')
```

```python
# Load the Corpus
from nltk.corpus import brown
```

```python
# View Corpus Information
print(brown.categories())
```

```python
# Access Words from Corpus
words = brown.words()
print(words[:20])
```

```python
# Access Sentences
sentences = brown.sents()
print(sentences[0])
```

```python
# Access Category-Specific Corpus
news_words = brown.words(categories='news')
print(news_words[:20])
```

```python
# Corpus Size
print("Total words:", len(brown.words()))
print("Total sentences:", len(brown.sents()))
```

```python
# Word Frequency from Corpus
from nltk import FreqDist

freq = FreqDist(brown.words())
freq.most_common(10)
```

```python
# Custom Corpus (Your Own Text)
text = """NLP is amazing.
Corpus is the backbone of NLP."""

tokens = nltk.word_tokenize(text)
tokens
```
