Quantifiers




In Regular Expressions (Regex), quantifiers define how many times a character, group, or token should appear in the matching pattern.

They control the repetition of elements in the pattern.

```python
#Quantifiers *'?': Zero or one (tells whether a pattern is absent or present) print(find_pattern("ac", "ab?")) print(find_pattern("abc", "ab?")) print(find_pattern("abbc", "ab?")) [] +': One or more print(find_pattern("ac", "ab+")) print(find_pattern("abc", "ab+")) print(find_pattern("abbc", "ab+")) [] (n): Matches if a character is present exactly n number of times print(find_pattern("abbc", "ab{2}"))
```

```python
# "?" : zero or one ("tells whether a pattern is absent or present ") span	Meaning,(0,1) only index 0, (0,2) index 0 and 1, (1,3) index 1 and 2

print(find_pattern('ac','ab?'))   # only a found
print(find_pattern('abc','ab?'))  # both
print(find_pattern('acbb','ab?'))    # text , pattern
```

in the above example

 we have pattern  ab ? it is trying to fing out the "b" just after
  "a"



 String: "abbc"
Pattern: "ab?"

The first 'ab' is matched

The second 'b' is not included because ? allows only 0 or 1 b

```python
# "*" : zero or more ("tells whether a pattern is absent or present  and more than one  ")

print(find_pattern('ac','ab*'))   # only a found
print(find_pattern('abc','ab*'))  # both
print(find_pattern('abbc','ab*'))    # text , pattern    mutliple times
```

```python
# "+" one or more
print(find_pattern('ac','ab+'))   # only a found
print(find_pattern('abc','ab+'))
print(find_pattern('abbc','ab+'))
```

```python
# {'n'} : Matches if a character is present exactly  n muber of times
print(find_pattern('abbbc','ab{2}'))  #  n=2
print(find_pattern('abbbc','ab{4}'))  #  n=4
```

```python
#{m,n} :  Matches if a character is present  from m to n number of times
#  here m is starting position and n is ending position


print(find_pattern('aabbbbbbc','ab{3,5}'))   # retrun  true if 'b' is present 3-5 times
print(find_pattern('aabbbbbbc','ab{7,10}')) #7-10 times
print(find_pattern('aabbbbbbc','ab{,10}'))  # almost 10 times
print(find_pattern('aabbbbbbc','ab{10,}')) # at least 10 times
#
```

```python
print(find_pattern('a','^a'))  # starts with a
```

| **Regex Pattern** | **Meaning**                          | **Example Matches** | **Does NOT Match** |
| ----------------- | ------------------------------------ | ------------------- | ------------------ |
| `[abc]`           | Match **only a, b, or c**            | a, b, c             | x, p, 7            |
| `[abcabc]`        | Same as `[abc]` (duplicates ignored) | a, b, c             | x, p, 7            |
| `[a-z]`           | Match any **lowercase** letter       | a, m, z             | A, 5, @            |
| `[A-Z]`           | Match any **uppercase** letter       | A, M, Z             | a, 5, #            |
| `[a-zA-Z]`        | Match **any alphabet**               | a, Z, m, P          | 5, @, _            |
| `[0-9]`           | Match any **digit**                  | 0, 5, 9             | a, Z, !            |

Meta   Sequences






What Are Meta Sequences in Regex?

Meta sequences (or special sequences) in Regex are short notations that help match specific types of characters without writing long character sets manually.

They act like shortcuts to match:
digits
 letters
 whitespace
 boundaries
 any character (wildcard)


| **Pattern** | **Meaning**                       | **Example Matches**                          |
| ----------- | --------------------------------- | -------------------------------------------- |
| `\d`        | Digit                             | `0`, `9`, `5`                                |
| `\D`        | NOT digit                         | `a`, `@`, `X`                                |
| `\w`        | Word character (a-z, A-Z, 0-9, _) | `A`, `7`, `_`                                |
| `\W`        | NOT word character                | `@`, `%`, `#`                                |
| `\s`        | Whitespace (space, tab, newline)  | `' '`, `\t`, `\n`                            |
| `\S`        | NOT whitespace                    | `A`, `9`, `%`                                |
| `.`         | Any character except newline      | `a`, `@`, `5`                                |
| `\t`        | Tab                               | (tab character)                              |
| `\n`        | Newline                           | (line break)                                 |
| `\b`        | Word boundary                     | `\bcat` → matches `"cat"` in `"cat is here"` |
| `\B`        | NOT word boundary                 | Inside a word → e.g., `"scattered"`          |


The five most important re functions that you would be required to use most of the times are



match() Determine if the RE matches at the beginning of the string



search() Scan through a string, looking for any location where this RE matches



finall() Find all the substrings where the RE matches, and return them as a list



finditer() Find all substrings where RE matches and return them as asn iterator



sub() Find all substrings where the RE matches and substitute them with the given  string.

```python
# creating a function for match and comparing how differnt it is from re.search,re.match() Checks only at the beginning of the string

def match_pattern(text,patterns):
  if re.match(patterns,text):
    return re.match(patterns,text)
  else:
    return ('Not found')

```

```python
def find_pattern(text,pattern):    # creating a userdefined function for resuibility, re.search() Checks anywhere in the string
  if re.search(pattern,text):
    return re.search(pattern,text)
  else:
    return 'Not found'
#
```

```python
print(find_pattern('abbc','b+')) # b should be present at least 1 or more
print(match_pattern('abbc','b+'))#   the text is not starting with
print(match_pattern('bbc','b+'))#   the text is  starting with b
```

```python

url = 'https://example.com/2025/11/24/article-title'
date_regex = r'(\d{4})/(\d{1,2})/(\d{1,2})'   # No slashes at start & end  (\d{4}) → captures year,(\d{1,2}) → captures month,(\d{1,2}) → captures day

print(re.findall(date_regex, url))
```

| **Regex Part** | **Meaning**                                    | **Example Matches** |
| -------------- | ---------------------------------------------- | ------------------- |
| `r''`          | Raw string in Python (treats `\` literally)    | `r'\d{4}'`, `r'\s'` |
| `(\d{4})`      | Matches **4 digits** → captures **year**       | `2025`, `1999`      |
| `/`            | Matches a **literal slash**                    | `/`                 |
| `(\d{1,2})`    | Matches **1 or 2 digits** → captures **month** | `1`, `05`, `12`     |
| `/`            | Matches another **literal slash**              | `/`                 |
| `(\d{1,2})`    | Matches **1 or 2 digits** → captures **day**   | `7`, `09`, `31`     |
