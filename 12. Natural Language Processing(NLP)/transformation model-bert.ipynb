# BERT (Bidirectional Encoder Representations from Transformers) in full teaching style, split clearly into:

# What is BERT?

BERT stands for

Bidirectional Encoder Representations from Transformers

It is a Transformer-based NLP model developed by Google (2018).

📌 BERT understands context from both left and right sides of a word.

# Why BERT was needed?

Before BERT:

Models read text only left → right OR right → left

Context understanding was weak

# Key Feature of BERT ⭐
* Bidirectional Context
* LEFT ← word → RIGHT
* I love playing football

# BERT Architecture
* Input → Encoder → Contextual Embeddings

# Tasks BERT is good at

Text classification

Sentiment analysis

Named Entity Recognition (NER)

Question answering

Sentence similarity

# BERT INPUT REPRESENTATION (VERY IMPORTANT)

Each input token embedding is a sum of 3 embeddings:

1️⃣ Token Embedding

Actual word representation

2️⃣ Segment Embedding

Used when there are two sentences

3️⃣ Position Embedding

Gives word position (order)

# Masked Language Model (MLM)

BERT predicts the missing word

# Next Sentence Prediction (NSP)

```python
from transformers import BertTokenizer, BertModel  # transformers → Hugging Face library for pretrained NLP models,BertTokenizer → converts text → tokens (numbers),BertModel → pretrained BERT neural network
import torch  # torch → PyTorch (BERT runs on PyTorch)
```

```python
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")  # BERT tokenizer that:Splits text into sub-words,Converts words → token IDs (numbers),Uses the same vocabulary BERT was trained on
model = BertModel.from_pretrained("bert-base-uncased")  # pretrained BERT model weights:12 Transformer layers,Self-attention mechanism,Learned from millions of sentences
```

```python
text = "I love natural language processing"
```

```python
inputs = tokenizer(
    text,
    return_tensors="pt",  # Converts token IDs into PyTorch tensors, Required because BERT model expects tensors
    padding=True,  # BERT processes batches of sentences
    truncation=True  # BERT has a maximum input length (usually 512 tokens)
)
```

```python
with torch.no_grad():
    outputs = model(**inputs)
```

```python
outputs
```

```python
token_embeddings = outputs.last_hidden_state  # provides contextualized embeddings for each token from the final BERT layer
print(token_embeddings.shape)  # Batch size,Number of tokens,Embedding size (features)
```

```python
sentence_embedding = outputs.last_hidden_state[:, 0, :]  # batch_size, sequence_length, hidden_size, all sentences in the batch,the first token ([CLS]),all 768 hidden features
print(sentence_embedding.shape)
```

```python
from transformers import BertForSequenceClassification

model = BertForSequenceClassification.from_pretrained(
    "bert-base-uncased",
    num_labels=2
)
```

```python
inputs = tokenizer("I love this movie", return_tensors="pt")  # {'input_ids': tensor([[ 101, 1045, 2293, 2023, 3185, 102 ]]),'attention_mask': tensor([[1, 1, 1, 1, 1, 1]])}, # Return the tokenized output as PyTorch tensors
outputs = model(**inputs)  # Tokens → BERT embeddings

logits = outputs.logits  # Raw, unnormalized scores,Not probabilities,Higher = more confident
prediction = torch.argmax(logits, dim=1)  # torch is the main Python module of PyTorch

print(prediction)
```
