# Chatbot Architecture (Detailed Explanation)

Chatbot architecture defines how a chatbot understands user input and generates responses.

## 1. User Interface Layer
Purpose

Allows users to interact with the chatbot.

Examples

Web chat window

Mobile app

WhatsApp / Telegram bot

Voice (Alexa, Google Assistant)

Input types

Text

Voice (converted to text via ASR)

## 2. Input Processing (NLP Preprocessing)
Purpose

Clean and normalize raw user input.

Steps involved

Lowercasing

Removing punctuation

Tokenization

Stopword removal

Stemming / Lemmatization

Example
"HELLO!!! How are YOU?"
↓
* ["hello", "how", "are", "you"]


## 3. Intent Recognition
Purpose

Identify what the user wants.

Techniques
Rule-based

Keyword matching

Regex patterns

ML-based

TF-IDF + Naive Bayes

SVM

Deep Learning

LSTM / GRU

Transformers (BERT)

Example
"Book a flight"
* → Intent = booking


## 4. Entity Recognition (NER)
Purpose

Extract important information from the sentence.

Example
"Book a flight from Delhi to Mumbai tomorrow"

## 5. Dialogue Manager (Brain of Chatbot ⭐)
Purpose

Decides what to do next based on:

Current intent

Conversation history

Context

Responsibilities

Maintain conversation state

Handle multi-turn conversations

Decide next action

Example
* User: Book a flight
* Bot: From where?
* User: Delhi

## 6. Business Logic / Backend Integration
Purpose

Perform actual tasks.

Examples

Database queries

API calls

Payment processing

Order tracking

Example
Check order status → call order API

## 7. Response Generation
Purpose

Generate a response to the user.

Types
Rule-based
"Your order is confirmed."

Template-based
"Your order from {from} to {to} is booked."

NLG (Advanced)

GPT

Seq2Seq models

## 8. Output Processing
Purpose

Make response user-friendly.

Tasks

Text formatting

Emojis

Language translation

Text-to-speech (for voice bots)

## 9. Learning & Feedback Loop (Optional)
Purpose

Improve chatbot over time.

Includes

Logging conversations

User feedback

Model retraining
